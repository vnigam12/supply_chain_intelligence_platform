# Process data for relational database

```python
# Create the Dimension: Customers
dim_customer = df[[
    'Customer Id', 'Customer Fname', 'Customer Lname', 'Customer Segment',
    'Customer Email', 'Customer Password', 'Customer Street', 'Customer City',
    'Customer State', 'Customer Country', 'Customer Zipcode'
]].drop_duplicates(subset=['Customer Id'])
dim_customer['Customer Name'] = dim_customer['Customer Fname'] + ' ' + dim_customer['Customer Lname']
dim_customer = dim_customer.drop(columns=['Customer Fname', 'Customer Lname'])

# Preview the result
display(dim_customer)
```

```python
# Create the Dimension: Products
dim_product = df[[
    'Product Card Id', 'Product Name', 'Product Category Id', 'Product Description', 'Product Image', 'Product Price', 'Product Status'
]].drop_duplicates(subset=['Product Card Id'])

# Preview the result
display(dim_product)
```

```python
# Create the Dimension: Category dimension
dim_category = df[[
    'Category Id', 'Category Name' # Product category code and name
    ]].drop_duplicates()
```

```python
# Create the Dimension: Department dimension
dim_department = df[[
    'Department Id', 'Department Name' # Department code and name of store
    ]].drop_duplicates()
```

```python
# Create the Dimension: Location (Order Geographical) dimension
dim_location = df[['Order City', 'Order Country', 'Order Region', 'Order State', 'Market']].drop_duplicates().reset_index(drop=True)
dim_location['Location_Id'] = range(1, len(dim_location) + 1)

# Merge Location_Id back to df for the fact table
df = df.merge(dim_location, on=['Order City', 'Order Country', 'Order Region', 'Order State', 'Market',], how='inner')

# Preview the result
display(dim_location)

# Inspect one ID
filtered_data = dim_location[dim_location['Location_Id'] == 1452]
display(filtered_data)
```

```python
# Create the Dimension: Date dimension
df['order date (DateOrders)'] = pd.to_datetime(df['order date (DateOrders)'])
df['shipping date (DateOrders)'] = pd.to_datetime(df['shipping date (DateOrders)'])

# Combine all unique dates from both order and shipping dates to build a complete calendar
all_unique_dates = pd.concat([
    df['order date (DateOrders)'], 
    df['shipping date (DateOrders)']
]).unique()

dim_date = pd.DataFrame({'Date': all_unique_dates})

# Generate surrogate keys and date attributes
dim_date['Date_Key'] = dim_date['Date'].dt.strftime('%Y%m%d').astype(int)
dim_date['Year'] = dim_date['Date'].dt.year
dim_date['Month'] = dim_date['Date'].dt.month
dim_date['Day'] = dim_date['Date'].dt.day
dim_date['Quarter'] = dim_date['Date'].dt.quarter
dim_date['Weekday'] = dim_date['Date'].dt.day_name()

# Drop duplicates just to ensure complete integrity
dim_date = dim_date.drop_duplicates(subset=['Date_Key'])
```

```python
# Add Date_Key to df
df['Order_Date_Key'] = df['order date (DateOrders)'].dt.strftime('%Y%m%d').astype(int)
df['Shipping_Date_Key'] = df['shipping date (DateOrders)'].dt.strftime('%Y%m%d').astype(int)

# Add the Weekday to df
df['Order_Weekday'] = df['order date (DateOrders)'].dt.day_name()
```

```python
# Create the Dimension: Order details dimension
dim_order_details = df[[
    'Order Id', 'Order Customer Id', 'Order Status', 'Type', 'Order Item Cardprod Id', 'order date (DateOrders)', 'Order_Date_Key'
]].drop_duplicates(subset=['Order Id'])

# Preview the result
display(dim_order_details)
```

```python
# Create the Dimension: Shipping dimension
dim_shipping = df[[
    'Shipping Mode', 'Delivery Status', 
    'Late_delivery_risk', 'Days for shipping (real)', 'Days for shipment (scheduled)', 'shipping date (DateOrders)', 'Shipping_Date_Key'
]].drop_duplicates().reset_index(drop=True)

dim_shipping['Shipping_Id'] = range(1, len(dim_shipping) + 1)

# Merge back to df for the fact table
df = df.merge(dim_shipping, on=['shipping date (DateOrders)', 'Shipping_Date_Key', 'Shipping Mode', 'Delivery Status', 'Late_delivery_risk', 'Days for shipping (real)', 'Days for shipment (scheduled)'], how='inner')

# Preview the result
display(dim_shipping)
```

```python
# Create the Fact: Sales
fact_sales = df[[
    'Order Item Id', 'Order Id', 'Customer Id', 'Customer Segment', 'Product Card Id', 'Category Id', 
    'Department Id', 'Location_Id', 'Shipping_Id', 'Order_Date_Key', 'Shipping_Date_Key', 
    'Sales', 'Order Item Total', 'Order Profit Per Order', 'Benefit per order', 'Sales per customer',	
    'Order Item Quantity', 'Order Item Discount', 'Order_Weekday',
    'Order Item Discount Rate', 'Order Item Product Price', 'Order Item Profit Ratio',
    'Days for shipping (real)', 'Days for shipment (scheduled)', 'Order Status',
    'Latitude', 'Longitude', 'Order Region', 'Order Country', 'Customer Country'
]]

# Preview the result
display(fact_sales)
```

```python
# Check for duplicates in the primary key column
duplicates = fact_sales[fact_sales.duplicated(subset=['Order Item Id'])]
display(duplicates)
```

```python
# Standardize helper function (convert everything to lowercase)
def standardize_series(series):
    return series.astype(str).str.lower().str.strip()

# Get unique values to run matching once
unique_log_products = standardize_series(log['Product']).unique()
unique_log_cats = standardize_series(log['Category']).unique()
unique_log_depts = standardize_series(log['Department']).unique()

# Create fuzzy mapping dictionary function
def get_fuzzy_map(source_unique, target_df, target_col_name, id_col_name, threshold=85):
    target_names = standardize_series(target_df[target_col_name]).tolist()
    mapping = {}
    for val in source_unique:
        match, score = process.extractOne(val, target_names)
        if score >= threshold:
            # Get the original ID from the dimension table
            match_id = target_df.loc[standardize_series(target_df[target_col_name]) == match, id_col_name].iloc[0]
            mapping[val] = match_id
    return mapping

# 4. Generate the maps
prod_map = get_fuzzy_map(unique_log_products, dim_product, 'Product Name', 'Product Card Id')
cat_map = get_fuzzy_map(unique_log_cats, dim_category, 'Category Name', 'Category Id')
dept_map = get_fuzzy_map(unique_log_depts, dim_department, 'Department Name', 'Department Id')

# Apply mapping to the original log dataframe
log['Product_Card_Id'] = standardize_series(log['Product']).map(prod_map)
log['Category_Id'] = standardize_series(log['Category']).map(cat_map)
log['Department_Id'] = standardize_series(log['Department']).map(dept_map)

# Add Date Key
log['Date_Key'] = pd.to_datetime(log['Timestamp']).dt.strftime('%Y%m%d').astype(int)
```

```python
# Drop logs with no product match
log = log.dropna(subset=['Product_Card_Id'])

# Sort logs and orders by Product and Time (Time must be the primary sort key for merge_asof with 'by')
log = log.sort_values('Timestamp')
log['Product_Card_Id'] = log['Product_Card_Id'].astype(int)
order_data = df[['Order Id', 'Product Card Id', 'order date (DateOrders)']].drop_duplicates()
order_data = order_data.rename(columns={'Product Card Id': 'Product_Card_Id', 'order date (DateOrders)': 'Order_Time'})
order_data = order_data.sort_values('Order_Time')
```

```python
# Use merge_asof for proximity matching
# This finds the nearest order_time >= timestamp for each log entry within the same Product_Card_Id
linked_logs = pd.merge_asof(
    log, 
    order_data,
    left_on='Timestamp', 
    right_on='Order_Time',
    by='Product_Card_Id',
    direction='forward', # Look for orders AFTER the log
    tolerance=pd.Timedelta('24 hours') # Only link if order is within 24 hours
)
```

```python
# Fact table: Web traffic
fact_web_traffic = linked_logs[[
    'Product_Card_Id', 'Category_Id', 'Department_Id', 'Date_Key', 
    'ip', 'url', 'Timestamp', 'Order Id'
]].rename(columns={'ip': 'IP_Address', 'url': 'URL', 'Order Id': 'Associated_Order_Id'})

fact_web_traffic['Log_Id'] = range(1, len(fact_web_traffic) + 1)

# Preview the result
display(fact_web_traffic)
```

```python
# Show count of non-NaN vs NaN in Associated_Order_Id
order_id_counts = fact_web_traffic['Associated_Order_Id'].value_counts(dropna=False)

# Display the result
print("Count of records with Order IDs vs. those without (NaN):")
print(order_id_counts)
```
